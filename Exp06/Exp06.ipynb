```python
import pandas as pd
path='/content/drive/My Drive/ML Lab/Exp07/MumbaiMatch.csv'
df=pd.read_csv(path)
print(df)
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder(sparse_output=False)
encoded=encoder.fit_transform(df[['Toss','Temp','Venue']])
encoded_df=pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(['Toss','Temp','Venue'])
)
# Remove original categorical columns
df = df.drop(columns=['Toss','Temp','Venue'])
# Add encoded columns to existing df
df = pd.concat([df, encoded_df], axis=1)
print(df)

x=df.drop(columns=['Match'])
y=df['Match']

#Prior Probability
wins=4
losses=3
prob_win=4/7
prob_loss=3/7
```

```python
new_toss = 'L'
new_temp = 'COOL'
new_venue = 'JAIPUR'

# Likelihood for Win
win_data = df[df['Match'] == 'W']
p_toss_w_win = (win_data[f'Toss_{new_toss}'] == 1.0).sum() / wins
p_temp_hot_win = (win_data[f'Temp_{new_temp}'] == 1.0).sum() / wins
p_venue_mumbai_win = (win_data[f'Venue_{new_venue}'] == 1.0).sum() / wins

win_score=prob_win*p_toss_w_win*p_temp_hot_win*p_venue_mumbai_win

# Likelihood for Loss
loss_data = df[df['Match'] == 'L']
p_toss_w_loss = (loss_data[f'Toss_{new_toss}'] == 1.0).sum() / losses
p_temp_hot_loss = (loss_data[f'Temp_{new_temp}'] == 1.0).sum() / losses
p_venue_mumbai_loss = (loss_data[f'Venue_{new_venue}'] == 1.0).sum() / losses

loss_score=prob_loss*p_toss_w_loss*p_temp_hot_loss*p_venue_mumbai_loss

if win_score > loss_score:
    print("Prediction:WIN")
else:
    print("Prediction:LOSS")
print(loss_score)
```
