Drop Unneceary Features

```python
df=df.drop(columns=['YEAR','HOST','MATCHES PLAYED','GOALS SCORED'])
```

Handling Non-Numeric Data

```python
from sklearn.preprocessing import OneHotEncoder

encoder=OneHotEncoder(sparse_output=False)
encoded=encoder.fit_transform(df[['CHAMPION','RUNNER UP','THIRD PLACE']])
encoded_df=pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(['CHAMPION','RUNNER UP','THIRD PLACE'])
)
```

```python
col = 'TEAMS'
df[col] = (df[col] - df[col].min()) / (df[col].max() - df[col].min())
df.head(22)
```
